```python
)def generate_random_sub_networks(main_network: nx.Graph, nodes_in_sub_network:int) -> list[list[int]]:
    if nodes_in_sub_network % 2:
        print("Subgraphs have unequal lengths!")
        return []
    nodes = list(main_network)
    random.shuffle(nodes)
    if len(nodes) % nodes_in_sub_network != 0:
        print("Subgraphs have unequal lengths!")
        return[]
    nsub_networks = int(len(nodes) / nodes_in_sub_network)

    return [nodes[i*nodes_in_sub_network : (i+1)*nodes_in_sub_network] for i in range(nsub_networks)]
```

```python
networks = generate_random_sub_networks(main_network, 6)
```

```python
sub_networks = [main_network.subgraph(i) for i in networks]
```

```python
nx.draw(sub_networks[0], with_labels=True)
```
